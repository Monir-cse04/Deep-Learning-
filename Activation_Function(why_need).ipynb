```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_moons

# Generate 2D binary classification dataset with a moon shape
X, y = make_moons(n_samples=500, noise=0.1, random_state=42)
```

```python
# Create a DataFrame for visualization
df = pd.DataFrame(dict(x=X[:,0], y=X[:,1], label=y))
```

```python
# Plot the dataset
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='x', y='y', hue='label', palette='viridis')
plt.title(' Moon Dataset')
plt.xlabel('class 1')
plt.ylabel('class 2')
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(128, input_dim=2, activation='linear')) # used linaer(no AF) in hidden layer
model.add(Dense(128, activation='linear')) # linear means no AF is used
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history = model.fit(X, y, epochs=100, batch_size=32, validation_split=0.2,verbose=True)
```

```python
from mlxtend.plotting import plot_decision_regions
```

# without Activation function in non linear dataset

```python
plot_decision_regions(X, y, clf=model)
plt.title('Moon Dataset(cant seperate dataset)')
plt.xlabel('class 1')
plt.ylabel('class 2')
plt.show()
```

> jokhon AF used na kora hoy tokhon ANN behave like a linear or logitics linear regression . like above

# apply AF in hidden layer

```python
model = Sequential()
model.add(Dense(128, input_dim=2, activation='relu')) # used AF in hidden layer
model.add(Dense(128, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history = model.fit(X, y, epochs=100, batch_size=32, validation_split=0.2,verbose=True)
```

# activation function for non linear

```python
plot_decision_regions(X, y, clf=model)
plt.title('Moon Dataset with applying AF')
plt.xlabel('class 1')
plt.ylabel('class 2')
plt.show()
```
