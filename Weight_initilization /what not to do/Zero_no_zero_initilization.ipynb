```python
df = pd.read_csv('/content/drive/MyDrive/DL/dataset/ushape.csv', header=None, names=['X', 'Y', 'class'])
```

```python
plt.scatter(df['X'], df['Y'], c=df['class'], cmap='coolwarm')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('U-Shape Dataset')
plt.show()
```

```python
X= df.drop('class', axis=1).values
y= df['class'].values
```

```python
y
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

# for sigmoid

```python
model = Sequential()

model.add(Dense(2,activation='sigmoid',input_dim=2))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
# Set parameters to 0
model.get_weights()
```

```python
initial_weights = model.get_weights()
```

```python
initial_weights[0] = np.zeros(model.get_weights()[0].shape)
initial_weights[1] = np.zeros(model.get_weights()[1].shape)
initial_weights[2] = np.zeros(model.get_weights()[2].shape)
initial_weights[3] = np.zeros(model.get_weights()[3].shape)
```

```python
model.set_weights(initial_weights)
```

```python
model.get_weights()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model.fit(X,y,epochs=100,validation_split=0.2)
```

```python
model.get_weights()
```

# first 2 weight and 2nd two weight are also same

> acting as a single node. made linear

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype('int'), clf=model, legend=2)
```

# so here we see that no training take palce . when w =0 and b=0
